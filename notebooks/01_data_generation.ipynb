## Import Library & Configuration

## Periode Data

```python
dates = pd.date_range(
    start="2019-01-01",
    end="2025-12-01",
    freq="MS"
)

print("Start:", dates.min())
print("End  :", dates.max())
print("Number of months:", len(dates))

# 7 tahun × 12 bulan = 84 bulan
```

## Master Estate

```python
estate_data = {
    "estate_id": [
        "EST001", "EST002", "EST003", "EST004", "EST005",
        "EST006", "EST007", "EST008", "EST009", "EST010"
    ],

    "estate_name": [
        "Estate Lampung 01",
        "Estate Lampung 02",
        "Estate Sumsel 01",
        "Estate Jambi 01",
        "Estate Riau 01",
        "Estate Kalbar 01",
        "Estate Kalteng 01",
        "Estate Kaltim 01",
        "Estate Sulawesi 01",
        "Estate Sulawesi 02"
    ],

    "region": [
        "Sumatra",
        "Sumatra",
        "Sumatra",
        "Sumatra",
        "Sumatra",
        "Kalimantan",
        "Kalimantan",
        "Kalimantan",
        "Sulawesi",
        "Sulawesi"
    ],

    "area_ha": [
        4250,
        3800,
        5100,
        4600,
        5500,
        4100,
        4700,
        3900,
        3600,
        4300
    ],

    "planting_year": [
        2010,
        2012,
        2009,
        2011,
        2008,
        2013,
        2010,
        2014,
        2011,
        2012
    ]
}

df_estate = pd.DataFrame(estate_data)

df_estate
```

## Karakteristik Produksi Estate

```python
production_capacity = {
    "EST001": 2.05,
    "EST002": 2.10,
    "EST003": 2.00,
    "EST004": 2.15,
    "EST005": 2.05,
    "EST006": 2.20,
    "EST007": 2.10,
    "EST008": 2.25,
    "EST009": 2.00,
    "EST010": 2.10
}

rainfall_base = {
    "Sumatra": 220,
    "Kalimantan": 250,
    "Sulawesi": 190
}

df_estate["base_yield_ton_ha"] = df_estate["estate_id"].map(
    production_capacity
)

df_estate["base_rainfall_mm"] = df_estate["region"].map(
    rainfall_base
)

df_estate
```

## Generate Kombinasi Estate × Month

```python
# setiap estate mempunyai observasi untuk setiap bulan.
df_base = pd.MultiIndex.from_product(
    [
        df_estate["estate_id"],
        dates
    ],
    names=["estate_id", "date"]
).to_frame(index=False)

df_base.head()
```

```python
print("Expected rows:", len(df_estate) * len(dates))
print("Actual rows  :", len(df_base))
```

## Gabung Informasi Estate

```python
df_base = df_base.merge(
    df_estate,
    on="estate_id",
    how="left"
)

df_base.head()
```

## hitung Plant Age

```python
df_base["plant_age_years"] = (
    df_base["date"].dt.year
    - df_base["planting_year"]
)

df_base[
    ["estate_id", "date", "planting_year", "plant_age_years"]
].head(10)
```

## buat Seasonality

```python
# Karena produksi perkebunan bisa memiliki pola musiman, kita buat seasonal factor berdasarkan bulan.
# Jadi nanti kita akan mempunyai pola:
seasonal_factor = {
    1: 0.95,
    2: 0.97,
    3: 1.02,
    4: 1.05,
    5: 1.08,
    6: 1.12,
    7: 1.15,
    8: 1.12,
    9: 1.05,
    10: 1.00,
    11: 0.96,
    12: 0.93
}

df_base["month"] = df_base["date"].dt.month

df_base["seasonal_factor"] = df_base["month"].map(
    seasonal_factor
)

df_base[
    ["date", "month", "seasonal_factor"]
].drop_duplicates().sort_values("date")
```

## Generate Rainfall

```python
# Kita menggunakan: regional rainfall baseline, seasonality, random variation
rainfall_seasonal = {
    1: 1.15,
    2: 1.10,
    3: 1.05,
    4: 0.95,
    5: 0.90,
    6: 0.85,
    7: 0.80,
    8: 0.82,
    9: 0.90,
    10: 1.00,
    11: 1.10,
    12: 1.18
}

df_base["rainfall_factor"] = df_base["month"].map(
    rainfall_seasonal
)

rainfall_noise = np.random.normal(
    loc=0,
    scale=25,
    size=len(df_base)
)

df_base["rainfall_mm"] = (
    df_base["base_rainfall_mm"]
    * df_base["rainfall_factor"]
    + rainfall_noise
)

df_base["rainfall_mm"] = (
    df_base["rainfall_mm"]
    .clip(lower=80)
    .round(1)
)
```

```python
df_base[
    ["estate_id", "date", "region", "rainfall_mm"]
].head(15)
```

## Generate Rainy Days

```python
# Jumlah rainy days harus berhubungan dengan rainfall.
rainy_day_noise = np.random.normal(
    loc=0,
    scale=1.5,
    size=len(df_base)
)

df_base["rainy_days"] = (
    df_base["rainfall_mm"] / 15
    + rainy_day_noise
).round()

df_base["rainy_days"] = (
    df_base["rainy_days"]
    .clip(lower=5, upper=28)
    .astype(int)
)
```

# Generate Fertilizer

```python
# Kita buat fertilizer berdasarkan luas area, sekitar 8–12 kg fertilizer / ha / month
fertilizer_rate = np.random.uniform(
    8,
    12,
    size=len(df_base)
)

fertilizer_noise = np.random.normal(
    0,
    0.05,
    size=len(df_base)
)

df_base["fertilizer_kg"] = (
    df_base["area_ha"]
    * fertilizer_rate
    * (1 + fertilizer_noise)
)

df_base["fertilizer_kg"] = (
    df_base["fertilizer_kg"]
    .clip(lower=0)
    .round(0)
)
```

## Generate Operational Working Days

```python
# Dalam satu bulan, effective working days kira-kira 22–27 hari.
df_base["working_days"] = np.random.randint(
    22,
    27,
    size=len(df_base)
)
```

## Generate Operational Disruptions

```python
# buat sebagian besar bulan memiliki disruption rendah.
df_base["disruption_days"] = np.random.poisson(
    lam=1.2,
    size=len(df_base)
)

df_base["disruption_days"] = (
    df_base["disruption_days"]
    .clip(lower=0, upper=7)
)
```

## buat Efek Umur Tanaman

```python
# Kita buat pola sederhana: tanaman muda → produksi meningkat, usia produktif → produksi optimal, usia tua → produktivitas mulai

def age_factor(age):
    if age < 6:
        return 0.75
    elif age < 10:
        return 0.90
    elif age <= 18:
        return 1.05
    elif age <= 22:
        return 0.95
    else:
        return 0.82

df_base["age_factor"] = df_base["plant_age_years"].apply(
    age_factor
)
```

## buat Efek Rainfall terhadap Produksi

```python
# Kita tidak ingin: semakin tinggi rainfall → produksi selalu semakin tinggi. Karena terlalu banyak hujan juga bisa mengganggu operasional. Kita buat rainfall optimal sekitar 180–280 mm.
def rainfall_factor(rainfall):
    if rainfall < 120:
        return 0.82
    elif rainfall < 180:
        return 0.92
    elif rainfall <= 280:
        return 1.05
    elif rainfall <= 350:
        return 0.98
    else:
        return 0.88

df_base["rainfall_production_factor"] = (
    df_base["rainfall_mm"].apply(rainfall_factor)
)
```

## Efek Fertilizer

```python
df_base["fertilizer_rate_kg_ha"] = (
    df_base["fertilizer_kg"]
    / df_base["area_ha"]
)
```

```python
def fertilizer_factor(rate):
    if rate < 8:
        return 0.90
    elif rate <= 11:
        return 1.00
    elif rate <= 13:
        return 1.04
    else:
        return 1.02

df_base["fertilizer_production_factor"] = (
    df_base["fertilizer_rate_kg_ha"]
    .apply(fertilizer_factor)
)
```

## Efek Operational Disruption

```python
# Semakin banyak disruption days, semakin rendah produksi.

df_base["operation_factor"] = (
    1 - (df_base["disruption_days"] * 0.025)
)

df_base["operation_factor"] = (
    df_base["operation_factor"]
    .clip(lower=0.80)
)
```

## Generate TBS Production

```python
# baseline: Area × Base Yield
random_noise = np.random.normal(
    loc=1.0,
    scale=0.05,
    size=len(df_base)
)

df_base["tbs_production_ton"] = (
    df_base["area_ha"]
    * df_base["base_yield_ton_ha"]
    * df_base["age_factor"]
    * df_base["rainfall_production_factor"]
    * df_base["fertilizer_production_factor"]
    * df_base["operation_factor"]
    * df_base["seasonal_factor"]
    * random_noise
)
```

```python
df_base["tbs_production_ton"] = (
    df_base["tbs_production_ton"]
    .clip(lower=0)
    .round(1)
)
```

```python
df_base[
    [
        "estate_id",
        "date",
        "area_ha",
        "plant_age_years",
        "rainfall_mm",
        "fertilizer_kg",
        "disruption_days",
        "tbs_production_ton"
    ]
].head(15)
```

## Generate CPO Production

```python
# buat CPO berdasarkan TBS, gunakan extraction rate sekitar 18–22%.
cpo_extraction_rate = np.random.normal(
    loc=0.20,
    scale=0.015,
    size=len(df_base)
)

cpo_extraction_rate = np.clip(
    cpo_extraction_rate,
    0.17,
    0.23
)

df_base["cpo_extraction_rate"] = cpo_extraction_rate

df_base["cpo_production_ton"] = (
    df_base["tbs_production_ton"]
    * df_base["cpo_extraction_rate"]
).round(1)
```

## Generate Kernel Production

```python
# gunakan extraction rate sekitar 4–6%.
kernel_extraction_rate = np.random.normal(
    loc=0.05,
    scale=0.005,
    size=len(df_base)
)

kernel_extraction_rate = np.clip(
    kernel_extraction_rate,
    0.04,
    0.06
)

df_base["kernel_extraction_rate"] = (
    kernel_extraction_rate
)

df_base["kernel_production_ton"] = (
    df_base["tbs_production_ton"]
    * df_base["kernel_extraction_rate"]
).round(1)
```

## Copy Dataset untuk Data Quality Simulation

```python
weather_raw = df_weather.copy()
operation_raw = df_operation.copy()
production_raw = df_production.copy()
```

## Inject Missing Values

```python
# weather
weather_missing_idx = np.random.choice(
    weather_raw.index,
    size=15,
    replace=False
)

weather_raw.loc[
    weather_missing_idx,
    "rainfall_mm"
] = np.nan
```

```python
# operation
operation_missing_idx = np.random.choice(
    operation_raw.index,
    size=10,
    replace=False
)

operation_raw.loc[
    operation_missing_idx,
    "fertilizer_kg"
] = np.nan
```

```python
# production
production_missing_idx = np.random.choice(
    production_raw.index,
    size=8,
    replace=False
)

production_raw.loc[
    production_missing_idx,
    "tbs_production_ton"
] = np.nan
```

## Inject Duplicate Records

```python
# weather duplicates
weather_duplicates = weather_raw.sample(
    5,
    random_state=42
)

weather_raw = pd.concat(
    [weather_raw, weather_duplicates],
    ignore_index=True
)
```

```python
# operation duplicates
operation_duplicates = operation_raw.sample(
    5,
    random_state=42
)

operation_raw = pd.concat(
    [operation_raw, operation_duplicates],
    ignore_index=True
)
```

```python
# production duplicates
production_duplicates = production_raw.sample(
    5,
    random_state=42
)

production_raw = pd.concat(
    [production_raw, production_duplicates],
    ignore_index=True
)
```

## Inject Invalid Rainfall

```python
invalid_rainfall_idx = np.random.choice(
    weather_raw.index,
    size=5,
    replace=False
)

weather_raw.loc[
    invalid_rainfall_idx,
    "rainfall_mm"
] = -50
```

## Inject Invalid Fertilizer

```python
invalid_fertilizer_idx = np.random.choice(
    operation_raw.index,
    size=5,
    replace=False
)

operation_raw.loc[
    invalid_fertilizer_idx,
    "fertilizer_kg"
] = -1000
```

## Inject Invalid Working Days

```python
invalid_working_idx = np.random.choice(
    operation_raw.index,
    size=5,
    replace=False
)

operation_raw.loc[
    invalid_working_idx,
    "working_days"
] = 45
```

## Inject Invalid Production

```python
invalid_production_idx = np.random.choice(
    production_raw.index,
    size=5,
    replace=False
)

production_raw.loc[
    invalid_production_idx,
    "tbs_production_ton"
] = -500
```

## Inject Outlier

```python
outlier_idx = np.random.choice(
    production_raw.index,
    size=5,
    replace=False
)

production_raw.loc[
    outlier_idx,
    "tbs_production_ton"
] *= 2.5

production_raw.loc[
    outlier_idx,
    "tbs_production_ton"
] = (
    production_raw.loc[
        outlier_idx,
        "tbs_production_ton"
    ].round(1)
)
```
